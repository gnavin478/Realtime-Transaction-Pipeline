```sql
SELECT
    dc.customer_segment,
    dc.is_first_purchase,
    COUNT(DISTINCT ft.event_id) AS total_transactions,
    SUM(ft.net_amount) AS total_revenue,
    SUM(ft.profit_amount) AS total_profit,
    ROUND(AVG(ft.net_amount), 2) AS avg_order_value
FROM fact_transactions ft
JOIN dim_customer dc
    ON ft.customer_id = dc.customer_id
GROUP BY dc.customer_segment, dc.is_first_purchase
ORDER BY total_revenue DESC;
```
