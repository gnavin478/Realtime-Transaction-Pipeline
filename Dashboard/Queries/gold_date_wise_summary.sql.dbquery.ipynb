```sql
SELECT
    dd.full_date,
    dd.year,
    dd.month,
    dd.quarter,
    dd.day_name,
    COUNT(DISTINCT ft.event_id) AS total_transactions,
    SUM(ft.net_amount) AS total_revenue,
    SUM(ft.profit_amount) AS total_profit
FROM fact_transactions ft
JOIN dim_date dd
    ON CAST(ft.event_time AS DATE) = dd.full_date
GROUP BY dd.full_date, dd.year, dd.month, dd.quarter, dd.day_name
ORDER BY dd.full_date;
```
