```sql
SELECT
    dl.city,
    dl.state,
    dl.country,
    COUNT(DISTINCT ft.event_id) AS total_transactions,
    SUM(ft.net_amount) AS total_revenue
FROM fact_transactions ft
JOIN dim_location dl
    ON ft.city = dl.city
   AND ft.state = dl.state
   AND ft.country = dl.country
   AND ft.zip_code = dl.zip_code
GROUP BY dl.city, dl.state, dl.country
ORDER BY total_revenue DESC;
```
