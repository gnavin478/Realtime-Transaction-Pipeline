```sql
-- Null Check
SELECT * FROM fact_transactions WHERE net_amount IS NULL OR event_time IS NULL;

--Duplicate Check
SELECT event_id, COUNT(*) FROM fact_transactions GROUP BY event_id HAVING COUNT(*) > 1;

--Negative/ Invalid Check
SELECT * FROM fact_transactions WHERE net_amount < 0 OR quantity <= 0;

--Revenue Check
SELECT * FROM fact_transactions
WHERE net_amount != (gross_amount - discount_amount + tax_amount);

--Integirity Check
SELECT ft.customer_id FROM fact_transactions ft LEFT JOIN dim_customer dc ON ft.customer_id = dc.customer_id
WHERE dc.customer_id IS NULL;

--Date Validity
SELECT * FROM fact_transactions WHERE event_time > current_timestamp();

```
